```python
# Render HTML and compile to PDF via headless Edge/Chrome
browser_exe = Path(r"C:\Program Files (x86)\Microsoft\Edge\Application\msedge.exe")
env = Environment(loader=FileSystemLoader(TEMPLATES_DIR))
temp_dir = OUTPUT_DIR / "_temp_html"
temp_dir.mkdir(parents=True, exist_ok=True)

for doc in DOCUMENTS:
    tmpl = env.get_template(doc["template"])
    html = tmpl.render(**doc["data"])
    html_file = temp_dir / f"{doc['doc_id']}.html"
    html_file.write_text(html, encoding="utf-8")
    
    pdf_file = OUTPUT_DIR / f"{doc['doc_id']}.pdf"
    cmd = [
        str(browser_exe),
        "--headless",
        "--disable-gpu",
        "--no-margins",
        f"--print-to-pdf={pdf_file.resolve()}",
        str(html_file.resolve())
    ]
    res = subprocess.run(cmd, capture_output=True, text=True)
    if res.returncode == 0:
        print(f"✅ Generated PDF: {pdf_file.name}")
    else:
        print(f"❌ Failed {doc['doc_id']}: {res.stderr}")
```
